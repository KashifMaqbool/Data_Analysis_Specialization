## Task - 1

### Evaluate the dataset for missing data
Missing data was last converted from '?' to numpy.NaN. Pandas uses NaN and Null values interchangeably. This means, you can just identify the entries having Null values. Write a code that identifies which columns have missing data. 


````python
missing_data = df.notnull()
print(missing_data.head())
````

````python
for column in missing_data.columns.values.tolist():
    print(missing_data[column].value_counts())
    print("")
````

## Task - 2

### Replace with mean
Missing values in attributes that have continuous data are best replaced using Mean value. We note that values in "Weight_kg" attribute are continuous in nature, and some values are missing. Therefore, write a code to replace the missing values of weight with the average value of the attribute.


````python
weight_avg = df['Weight_kg'].astype('float').mean(axis = 0)
print("The Average of Weight_kg column is:", weight_avg)
````

````python
df['Weight_kg'] = df['Weight_kg'].fillna(weight_avg)
````

````python
df['Weight_kg'].isnull().sum()
````

### Replace with the most frequent value
Missing values in attributes that have categorical data are best replaced using the most frequent value. We note that values in "Screen_Size_cm" attribute are categorical in nature, and some values are missing. Therefore, write a code to replace the missing values of Screen Size with the most frequent value of the attribute.


````python
df.head(2)
````

````python
screen_size_frequent_value = df['Screen_Size_inch'].value_counts().idxmax()
print("The most frequent value for Screen_Size_inch is: ", screen_size_frequent_value)
````

````python
df.fillna({'Screen_Size_inch':screen_size_frequent_value}, inplace = True)
````

````python
df.info()
````

As we have noticed that There is not a single null value insie the dataframe.

## Task - 3

### Fixing the data types
Both "Weight_kg" and "Screen_Size_cm" are seen to have the data type "Object", while both of them should be having a data type of "float". Write a code to fix the data type of these two columns.


````python
df[["Weight_kg","Screen_Size_inch"]] = df[["Weight_kg","Screen_Size_inch"]].astype("float")
print(df['Weight_kg'].dtype)
print(df['Screen_Size_inch'].dtype)
````

## Task - 4

### Data Standardization
The value of Screen_size usually has a standard unit of inches. Similarly, weight of the laptop is needed to be in pounds. Use the below mentioned units of conversion and write a code to modify the columns of the dataframe accordingly. Update their names as well.

```{math}
1 inch = 2.54 cm
1 kg   = 2.205 pounds
```


##### Screen_Size_inch data standardization from inches to centimeters

````python
df['Screen_Size_inch'] = df['Screen_Size_inch']/2.54
````

##### Weight_kg data standardization from kg to pounds

````python
df["Weight_kg"] = df["Weight_kg"]*2.205
df.rename(columns={'Weight_kg':'Weight_pounds'}, inplace=True)
````

````python
df.head()
````

### Data Normalization
Often it is required to normalize a continuous data attribute. Write a code to normalize the "CPU_frequency" attribute with respect to the maximum value available in the dataset.


````python
# Write your code below and press Shift+Enter to execute
df['CPU_frequency'] = df['CPU_frequency']/df['CPU_frequency'].max()
df.head()
````

## Task - 5

### Binning
Binning is a process of creating a categorical attribute which splits the values of a continuous data into a specified number of groups. In this case, write a code to create 3 bins for the attribute "Price". These bins would be named "Low", "Medium" and "High". The new attribute will be named "Price-binned".


````python
df['Price'].value_counts()
````

````python
df['Price'].dtype
````

````python
bins = np.linspace(min(df['Price']), max(df['Price']), 4)
bins
````

````python
group_names = ['Low', 'Medium', 'High']
````

````python
df['Price_groups'] = pd.cut(df['Price'], bins, labels = group_names, include_lowest = True)
df[['Price','Price_groups']].head()
````

````python
df.head()
````

### Also, plot the bar graph of these bins.


````python
df['Price_groups'].value_counts()
````

````python
%matplotlib inline
import matplotlib.pyplot as plt

plt.bar(group_names, df['Price_groups'].value_counts())

plt.xlabel('Price')
plt.ylabel('counts')
plt.title('Price_Binned')

plt.show()
````

## Task - 6

### Indicator variables
Convert the "Screen" attribute of the dataset into 2 indicator variables, "Screen-IPS_panel" and "Screen-Full_HD". Then drop the "Screen" attribute from the dataset.


````python
dummy_variable = pd.get_dummies(df['Screen'])
dummy_variable = dummy_variable.astype("int")
dummy_variable.head()
````

````python
dummy_variable.rename(columns = {'Full HD':'Screen-Full_HD', 'IPS Panel':'Screen-IPS_panel'}, inplace=True)
dummy_variable.head()
````

````python
df = pd.concat([df, dummy_variable], axis=1)
df.drop('Screen', axis = 1, inplace = True)
````

````python
df.head()
````

# Task 3 - GroupBy and Pivot Tables

Group the parameters "GPU", "CPU_core" and "Price" to make a pivot table and visualize this connection using the pcolor plot.


````python
# Create the group

df_gptest = df[['GPU', 'CPU_core', 'Price']]
grouped_test1 = df_gptest.groupby(['GPU','CPU_core'], as_index=False).mean()
print(grouped_test1)
````

````python
# Create the Pivot table

grouped_pivot = grouped_test1.pivot(index='GPU',columns='CPU_core')
print(grouped_pivot)
````

````python
# Create the Plot

fig, ax = plt.subplots()
im = ax.pcolor(grouped_pivot, cmap='RdBu')

# Label names
row_labels = grouped_pivot.columns.levels[1]
col_labels = grouped_pivot.index

# Move ticks and labels to the center
ax.set_xticks(np.arange(grouped_pivot.shape[1]) + 0.5, minor=False)
ax.set_yticks(np.arange(grouped_pivot.shape[0]) + 0.5, minor=False)

# Insert labels
ax.set_xticklabels(row_labels, minor=False)
ax.set_yticklabels(col_labels, minor=False)

fig.colorbar(im)
plt.show()
````

# Task 4 - Pearson Correlation and p-values

Use the `scipy.stats.pearsonr()` function to evaluate the Pearson Coefficient and the p-values for each parameter tested above. This will help you determine the parameters most likely to have a strong effect on the price of the laptops.


````python
for parameters in ['RAM_GB','CPU_frequency','Storage_GB_SSD','Screen_Size_inch','Weight_pounds','CPU_core','OS','GPU','Category']:
    pearson_coef, p_value = stats.pearsonr(df[parameters], df['Price'])
    print(parameters)
    print("The Pearson Correlation Coefficient for ",parameters,"is ",pearson_coef,"with a P-value of P=", p_value)
    print("")
````
